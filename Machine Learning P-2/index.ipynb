```python
from sklearn.linear_model import LinearRegression, LogisticRegression
import numpy as np
# Sample data
X = np.array([[1], [2], [3], [4], [5]])
y_continuous = np.array([2, 4, 6, 8, 10]) # Continuous target
y_binary = np.array([0, 0, 1, 1, 1]) # Binary target
# Linear Regression
lin_reg = LinearRegression()
lin_reg.fit(X, y_continuous)
print("Linear Regression Prediction:", lin_reg.predict([[8]]))
# Logistic Regression
log_reg = LogisticRegression()
log_reg.fit(X, y_binary)
print("Logistic Regression Probability:", log_reg.predict_proba([[3]]))
print("Logistic Regression Class:", log_reg.predict([[1]]))
```
