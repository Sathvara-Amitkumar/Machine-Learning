```python
x = df.drop(columns='price', axis=1)
y = df['price']
```

```python
df.columns
```

```python
df.info()
```

```python
x_one_encode = pd.get_dummies(x, columns=['model', 'transmission', 'fuelType'], drop_first=True)
x_one_encode = x_one_encode.astype('int')
```

```python
x_one_encode.sample(3)
```

```python
# Label Encoding

from sklearn.preprocessing import LabelEncoder

encode = LabelEncoder()
x_label_encode = x

columns = ['model', 'transmission', 'fuelType']

for i in columns:
    x_label_encode[i] = encode.fit_transform(x_label_encode[i])
```

```python
x_label_encode['model'].value_counts().sort_index()
```

```python
x_label_encode.head()
```

```python
# Applying Standard Scaler on OneHotEncoder

from sklearn.preprocessing import StandardScaler

scale = StandardScaler()
numeric_col = ['year','mileage','tax','mpg']

x_one_encode[numeric_col] = scale.fit_transform(x_one_encode[numeric_col])
```

```python
x_one_encode
```

```python
x_label_encode.columns
```

```python
# Applying Standard Scaler on LabelEncoder

x_label_encode[['model', 'year', 'transmission', 'mileage', 'fuelType', 'tax', 'mpg',
       'engineSize']] = scale.fit_transform(x_label_encode[['model', 'year', 'transmission', 'mileage', 'fuelType', 'tax', 'mpg',
       'engineSize']])
```

```python
x_label_encode
```

```python
# Training and Testing Data

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, Lasso, Ridge
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
```

# 1. Check for OneHotEncoder

```python
x_train, x_test, y_train, y_test = train_test_split(x_one_encode, y, test_size=0.20, random_state=42)
```

```python
model = LinearRegression()
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
y_pred, y_test
```

```python
r2 = r2_score(y_test, y_pred)
print("R² score =", r2)
```

```python
n = x_one_encode.shape[0]
p = x_one_encode.shape[1]

adjusted_r2 = 1 - ((1-r2) * (n-1) / (n - p - 1))
print("Adjusted R² =", adjusted_r2)
```

```python
print("MAE =", mean_absolute_error(y_test, y_pred))
print("MSE =", mean_squared_error(y_test, y_pred))
```

```python
print("Training Score:", model.score(x_train, y_train))
print("Testing Score:", model.score(x_test, y_test))
```

```python
lasso = Lasso(alpha=0.1)
lasso.fit(x_train, y_train)

lasso_pred = lasso.predict(x_test)
print(r2_score(y_test, lasso_pred))
```
