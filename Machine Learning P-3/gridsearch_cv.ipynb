```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings

warnings.filterwarnings("ignore")
```

```python
df = sns.load_dataset('iris')
```

```python
df.head()
```

```python
df['species'].unique()
```

```python
df.columns
```

```python
x = df.drop('species', axis=1)
y = df['species']
```

```python
x.shape, y.shape
```

```python
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC 
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.20, random_state=42)
```

```python
model_knn = KNeighborsClassifier()
model_knn.fit(x_train, y_train)
```

```python
model_knn.score(x_test, y_test)
```

```python
model_svm = SVC()
```

```python
model_svm.fit(x_train, y_train)
```

```python
model_svm.score(x_test, y_test)
```

# Grid Search CV

```python
from sklearn.model_selection import GridSearchCV
```

```python
classifier = GridSearchCV((model_svm), {
    'C':[1, 20, 30, 40],
    'kernel':['rbf', 'linear','sigmoid'],
    'gamma':['auto', 'scale']
}, cv=5, return_train_score=False)
```

```python
classifier.fit(x, y)
```

```python
classifier.cv_results_
```

```python
svm_data = pd.DataFrame(data=classifier.cv_results_)
```

```python
svm_data.head(2)
```

```python
svm_data[['param_C', 'param_kernel', 'mean_test_score']]
```

```python
# gridsearch cv for KNN

classifier2 = GridSearchCV((model_knn), {
    'n_neighbors': [5, 7, 9],
    'weights': ['uniform', 'distance'],
    'algorithm': ['auto', 'ball_tree', 'kd_tree'],
    'metric': ['minkowski', 'manhattan']
}, cv=5, return_train_score=False)
```

```python
classifier2.fit(x, y)
```

```python
knn_data = pd.DataFrame(classifier2.cv_results_)
```

```python
knn_data.head(2)
```

```python
knn_data[['param_n_neighbors','param_weights', 'param_metric', 'mean_test_score']].sort_values(by='mean_test_score', ascending=False)
```

# Random Search CV

```python
from sklearn.model_selection import RandomizedSearchCV

classifier_random = RandomizedSearchCV((model_knn), {
    'n_neighbors': [5, 7, 9],
    'weights': ['uniform', 'distance'],
    'algorithm': ['auto', 'ball_tree', 'kd_tree'],
    'metric': ['minkowski', 'manhattan']
}, n_iter=10, cv=5, return_train_score=False)
```

```python
classifier_random.fit(x, y)
```

```python
knn_random = pd.DataFrame(classifier_random.cv_results_)
```

```python
knn_random[['param_n_neighbors','param_weights', 'param_metric', 'mean_test_score']].sort_values(by='mean_test_score', ascending=False)
```
