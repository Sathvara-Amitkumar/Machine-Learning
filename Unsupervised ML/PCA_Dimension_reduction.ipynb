```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.datasets import make_blobs
```

```python
X, y = make_blobs(n_samples=600, n_features=6, centers=3, cluster_std=1.5, random_state=42)
```

```python
X
```

```python
scaler = StandardScaler()
x_scaled = scaler.fit_transform(X)
```

```python
x_scaled
```

```python
pca = PCA(n_components=2)
x_pca = pca.fit_transform(x_scaled)
```

```python
x_pca
```

```python
df_pca = pd.DataFrame(x_pca, columns=['PCA_1', 'PCA_2'])
df_pca['pca_label'] = y
```

```python
df_pca.columns
```

```python
df_pca.head()
```

```python
# Perform scatterplot to show the dimensions

sns.scatterplot(x='PCA_1', y='PCA_2', hue='pca_label', data=df_pca, palette='tab10')
```
